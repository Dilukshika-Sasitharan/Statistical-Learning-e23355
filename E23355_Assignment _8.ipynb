# Q1 – Statistical Mechanics of the Quantum Harmonic Oscillator

Consider a quantum harmonic oscillator of angular frequency **$\omega$** in thermal equilibrium at temperature **$T$**. The quantum mechanical energy eigenvalues are

$$
E_j=\hbar\omega\left(j+\frac{1}{2}\right), \qquad j=0,1,2,\ldots
$$

where

- **$\hbar$** is the reduced Planck constant.
- **$k_B$** is the Boltzmann constant.
- **$T$** is the absolute temperature.

Using the canonical ensemble, derive the partition function and the thermodynamic properties of the system.

---

## 1. Partition Function

The canonical partition function is

$$
Z(T)=\sum_{j=0}^{\infty}e^{-E_j/(k_BT)}.
$$

Substituting the energy eigenvalues

$$
E_j=\hbar\omega\left(j+\frac{1}{2}\right),
$$

gives

$$
Z(T)=
e^{-\frac{\hbar\omega}{2k_BT}}
\sum_{j=0}^{\infty}
\left(
e^{-\frac{\hbar\omega}{k_BT}}
\right)^j.
$$

The above summation is a geometric series with common ratio

$$
r=e^{-\frac{\hbar\omega}{k_BT}},
$$

where

$$
0<r<1.
$$

Hence,

$$
\sum_{j=0}^{\infty}r^j
=
\frac{1}{1-r}.
$$

Therefore,

$$
\boxed{
Z(T)=
\frac
{e^{-\frac{\hbar\omega}{2k_BT}}}
{1-e^{-\frac{\hbar\omega}{k_BT}}}
}
$$

Thus, the canonical partition function of the quantum harmonic oscillator is

$$
\boxed{
Z(T)=
\frac
{e^{-\frac{\hbar\omega}{2k_BT}}}
{1-e^{-\frac{\hbar\omega}{k_BT}}}
}
$$

This expression is obtained by recognizing the partition function as a convergent geometric series and forms the basis for deriving all thermodynamic properties of the quantum harmonic oscillator.

---



```python
import sympy as sp
from IPython.display import display

# Enable pretty mathematical output
sp.init_printing(use_unicode=True)

# Define symbols
hbar, omega, k_B, T, j = sp.symbols(
    'hbar omega k_B T j',
    positive=True
)

# Energy eigenvalues
E_j = hbar * omega * (j + sp.Rational(1, 2))

# Partition function
Z = sp.summation(
    sp.exp(-E_j / (k_B * T)),
    (j, 0, sp.oo)
)

print("Partition Function:")

display(sp.simplify(Z))
```

---
## 2. Helmholtz Free Energy

The Helmholtz free energy is defined by

$$
F(T)=-k_B T\ln Z,
$$

where \(Z\) is the canonical partition function.

Substituting

$$
Z(T)=
\frac{e^{-\frac{\hbar\omega}{2k_B T}}}
{1-e^{-\frac{\hbar\omega}{k_B T}}},
$$

we obtain

$$
F(T)
=
-k_B T
\ln
\left(
\frac
{e^{-\frac{\hbar\omega}{2k_B T}}}
{1-e^{-\frac{\hbar\omega}{k_B T}}}
\right).
$$

Using the logarithmic identity

$$
\ln\left(\frac{a}{b}\right)
=
\ln(a)-\ln(b),
$$

the above expression becomes

$$
F(T)
=
-k_B T
\left[
-\frac{\hbar\omega}{2k_B T}
-
\ln\left(1-e^{-\frac{\hbar\omega}{k_B T}}\right)
\right].
$$

Simplifying,

$$
\boxed{
F(T)
=
\frac{\hbar\omega}{2}
+
k_B T
\ln\left(
1-e^{-\frac{\hbar\omega}{k_B T}}
\right)
}
$$

The first term,

$$
\frac{\hbar\omega}{2},
$$

represents the **zero-point energy** of the quantum harmonic oscillator, while the second term represents the **temperature-dependent thermal contribution** arising from the thermal occupation of excited energy levels.

Thus, the Helmholtz free energy describes the maximum useful work that can be extracted from the system at constant temperature and volume.

---

```python
import sympy as sp
from IPython.display import display

# Initialize pretty printing
sp.init_printing(use_unicode=True)

# Define symbols
hbar, omega, beta = sp.symbols('hbar omega beta', positive=True)

# Partition function
Z = sp.exp(-hbar*omega*beta/2) / (1 - sp.exp(-hbar*omega*beta))

# Helmholtz Free Energy
F = -sp.log(Z) / beta

print("Helmholtz Free Energy:")

display(sp.simplify(F))
```

## 3. Internal Energy

The internal energy of the quantum harmonic oscillator is obtained from the partition function using the thermodynamic relation

$$
U
=
-
\frac{\partial \ln Z}{\partial \beta},
$$

where

$$
\beta
=
\frac{1}{k_B T}.
$$

The partition function is

$$
Z
=
\frac{e^{-\beta\hbar\omega/2}}
{1-e^{-\beta\hbar\omega}}.
$$

Taking the natural logarithm,

$$
\ln Z
=
-\frac{\beta\hbar\omega}{2}
-
\ln\left(1-e^{-\beta\hbar\omega}\right).
$$

Differentiating with respect to $\beta$,

$$
U
=
-
\frac{\partial \ln Z}{\partial \beta}
=
\hbar\omega
\left(
\frac{1}{2}
+
\frac{1}
{e^{\beta\hbar\omega}-1}
\right).
$$

Substituting

$$
\beta=\frac{1}{k_B T},
$$

the internal energy is

$$
\boxed{
U(T)
=
\hbar\omega
\left(
\frac{1}{2}
+
\frac{1}
{e^{\frac{\hbar\omega}{k_B T}}-1}
\right)
}
$$

The first term,

$$
\frac{\hbar\omega}{2},
$$

is the **zero-point energy**, which exists even at absolute zero temperature due to the quantum nature of the harmonic oscillator.

The second term,

$$
\frac{\hbar\omega}
{e^{\frac{\hbar\omega}{k_B T}}-1},
$$

represents the **thermal energy contribution** arising from the excitation of higher quantum energy states as the temperature increases.

Thus, the internal energy consists of both the temperature-independent zero-point energy and the temperature-dependent thermal excitation energy, providing the complete thermodynamic description of the quantum harmonic oscillator.

---


```python
import sympy as sp
from IPython.display import display

# Initialize pretty printing
sp.init_printing(use_unicode=True)

# Define symbols
hbar, omega, kB, T, beta = sp.symbols(
    'hbar omega k_B T beta',
    positive=True
)

# Partition function
Z_beta = sp.exp(-hbar * omega * beta / 2) / (
    1 - sp.exp(-hbar * omega * beta)
)

# Internal Energy
U = sp.simplify(-sp.diff(sp.log(Z_beta), beta))

# Express in terms of temperature
U_T = sp.simplify(
    U.subs(beta, 1 / (kB * T))
)

print("Internal Energy:")

display(U_T)
```

## 4. Entropy

The entropy of the quantum harmonic oscillator is obtained from the Helmholtz free energy using the thermodynamic relation

$$
S
=
-
\left(
\frac{\partial F}{\partial T}
\right)_V,
$$

where the Helmholtz free energy is

$$
F(T)
=
\frac{\hbar\omega}{2}
+
k_B T
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_B T}}
\right).
$$

Differentiating the Helmholtz free energy with respect to temperature gives

$$
\boxed{
S(T)
=
-k_B
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_B T}}
\right)
+
\frac{\hbar\omega}{T}
\frac{1}
{e^{\frac{\hbar\omega}{k_B T}}-1}
}
$$

The first term,

$$
-k_B
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_B T}}
\right),
$$

represents the **statistical entropy**, which arises from the distribution of the oscillator among the available quantum energy states.

The second term,

$$
\frac{\hbar\omega}{T}
\frac{1}
{e^{\frac{\hbar\omega}{k_B T}}-1},
$$

represents the **thermal contribution to the entropy**, describing the increase in entropy due to the thermal excitation of higher energy states as the temperature increases.

The derived thermodynamic quantities satisfy the fundamental thermodynamic identity

$$
\boxed{
U = F + TS,
}
$$

which confirms the thermodynamic consistency of the derived expressions.

Thus, the entropy consists of both a statistical contribution associated with the occupation of quantum states and a thermal contribution arising from temperature-dependent excitations, providing a complete thermodynamic description of the quantum harmonic oscillator.

---

```python
import sympy as sp
from IPython.display import display

# Initialize pretty printing
sp.init_printing(use_unicode=True)

# Define symbols
hbar, omega, kB, T = sp.symbols(
    'hbar omega k_B T',
    positive=True
)

# Helmholtz Free Energy
F_T = (
    hbar*omega/2
    + kB*T*sp.log(
        1 - sp.exp(-hbar*omega/(kB*T))
    )
)

# Entropy
S = sp.simplify(-sp.diff(F_T, T))

print("Entropy:")
display(S)

# Internal Energy
U_T = sp.simplify(
    hbar*omega*(
        sp.Rational(1, 2)
        + 1/(sp.exp(hbar*omega/(kB*T)) - 1)
    )
)

print("\nVerification of U = F + T*S:")

display(sp.simplify(F_T + T*S - U_T))
```

## 5. Heat Capacity

The constant-volume heat capacity is defined as

$$
C_V
=
\left(
\frac{\partial U}{\partial T}
\right)_V.
$$

Using the expression for the internal energy,

$$
U(T)
=
\hbar\omega
\left(
\frac{1}{2}
+
\frac{1}
{e^{\frac{\hbar\omega}{k_B T}}-1}
\right),
$$

the heat capacity is obtained by differentiating the internal energy with respect to temperature:

$$
C_V
=
\frac{\partial U}{\partial T}
=
k_B
\left(
\frac{\hbar\omega}{k_B T}
\right)^2
\frac{e^{\frac{\hbar\omega}{k_B T}}}
{\left(e^{\frac{\hbar\omega}{k_B T}}-1\right)^2}.
$$

Using the hyperbolic identity

$$
\sinh(x)
=
\frac{e^x-e^{-x}}{2},
$$

the heat capacity can be written in the compact form

$$
\boxed{
C_V
=
k_B
\left(
\frac{\hbar\omega}{2k_B T}
\right)^2
\frac{1}
{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}
}
$$

### High-Temperature Limit

In the high-temperature limit,

$$
k_B T \gg \hbar\omega,
$$

we have

$$
\sinh(x)\approx x,
\qquad
x=\frac{\hbar\omega}{2k_B T}.
$$

Therefore,

$$
\boxed{
C_V
\rightarrow
k_B.
}
$$
