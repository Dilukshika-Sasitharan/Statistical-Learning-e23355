# Q2)Bertrand's Paradox

A chord of a circle is chosen "at random." The probability that its length exceeds the side of the inscribed equilateral triangle depends on how "random chord" is defined.

A chord is longer than the side of the inscribed equilateral triangle if and only if its distance from the center is less than half the radius (R/2).

## Model 1: Random Endpoints on the Circle

Two points are chosen uniformly at random on the circumference of the circle.

$$
P = \frac{1}{3}
$$



## Model 2: Random Radius with Uniform Midpoint

A random radius is chosen, and a point is selected uniformly along the radius as the midpoint of the chord.

$$
P = \frac{1}{2}
$$



## Model 3: Random Midpoint in the Circle (Area-based)

The midpoint of the chord is chosen uniformly at random inside the circle.

$$
P = \frac{\pi (R/2)^2}{\pi R^2} = \frac{1}{4}
$$

- Random endpoints → $\frac{1}{3}$
- Random radius midpoint → $\frac{1}{2}$
- Random midpoint in circle → $\frac{1}{4}$


Different mathematically reasonable definitions of a "random chord" produce different probabilities.  
This is known as Bertrand's Paradox.
